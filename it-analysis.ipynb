```python
def load_plugin(year_month, plugin):

    path = (
        DATA_ROOT
        / f"year_month={year_month}"
        / f"plugin={plugin}"
        / "part-000.parquet"
    )

    print("Loading:")
    print(path)

    df = pd.read_parquet(path)

    if "timestamp" in df.columns:
        df["timestamp"] = pd.to_datetime(
            df["timestamp"],
            utc=True,
            errors="coerce"
        )

    return df
```

```python
import pandas as pd
import ast

# ── 1. Load job metadata ──────────────────────────────────────────────────────
jobs = pd.read_parquet(
    "data/dataset=main/year_month=22-03/plugin=job_table/part-000.parquet"
)
jobs["start_time"] = pd.to_datetime(jobs["start_time"], utc=True)
jobs["end_time"]   = pd.to_datetime(jobs["end_time"],   utc=True)

# ── 2. Load per-node power from ipmi_pub ─────────────────────────────────────
ipmi = pd.read_parquet(
    "data/dataset=main/year_month=22-03/plugin=ipmi_pub/part-000.parquet"
)
ipmi = ipmi[ipmi["metric"] == "total_power"].copy()
ipmi["timestamp"] = pd.to_datetime(ipmi["timestamp"], utc=True)

# ── 3. Helper: parse node list from job record ────────────────────────────────
def parse_nodes(node_str):
    """
    job["nodes"] is a string like "['node001', 'node002']" or "node001,node002".
    Adjust this parser to match the actual format in your data.
    """
    if pd.isna(node_str):
        return []
    try:
        parsed = ast.literal_eval(node_str)
        if isinstance(parsed, list):
            return parsed
    except (ValueError, SyntaxError):
        pass
    return [n.strip() for n in str(node_str).split(",") if n.strip()]

# ── 4. Per-job energy using only the job's allocated nodes ───────────────────
def job_energy_per_node(job, ipmi_df):
    nodes = parse_nodes(job["nodes"])
    if not nodes:
        return None

    # Filter: time window + allocated nodes only
    mask = (
        (ipmi_df["timestamp"] >= job["start_time"]) &
        (ipmi_df["timestamp"] <  job["end_time"]) &
        (ipmi_df["node"].isin(nodes))
    )
    slice_ = ipmi_df.loc[mask, ["timestamp", "node", "value_numeric"]].copy()

    if slice_.empty:
        return None

    # Integrate per node (W → Wh), then sum across nodes
    total_wh = 0.0
    for node, grp in slice_.groupby("node"):
        grp = grp.sort_values("timestamp")
        dt_hours = grp["timestamp"].diff().dt.total_seconds().div(3600).fillna(0)
        total_wh += (grp["value_numeric"] * dt_hours).sum()

    return total_wh / 1000  # convert Wh → kWh

jobs["node_energy_kwh"] = jobs.apply(
    lambda row: job_energy_per_node(row, ipmi), axis=1
)

# ── 5. Optional: also keep facility-level estimate for comparison ─────────────
power = pd.read_parquet(
    "data/dataset=main/year_month=22-03/plugin=logics_pub/part-000.parquet"
)
power = power[
    (power["metric"] == "Tot_ict") &
    (power["panel"].astype(str) == "generals") &
    (power["device"].astype(str) == "pue")
].copy()
power["timestamp"] = pd.to_datetime(power["timestamp"], utc=True)

def facility_energy_prorated(job, power_df):
    """Prorate facility IT load by job's node share."""
    mask = (
        (power_df["timestamp"] >= job["start_time"]) &
        (power_df["timestamp"] <  job["end_time"])
    )
    slice_ = power_df.loc[mask].copy()
    if len(slice_) < 2:
        return None
    dt_hours = slice_["timestamp"].diff().dt.total_seconds().div(3600).fillna(0)
    total_facility_kwh = (slice_["value_numeric"] * dt_hours).sum()

    # Prorate: job nodes / total cluster nodes
    total_nodes = 980  # M100 has 980 nodes — adjust if needed
    node_count  = len(parse_nodes(job["nodes"])) or 1
    return total_facility_kwh * (node_count / total_nodes)

jobs["facility_energy_kwh_prorated"] = jobs.apply(
    lambda row: facility_energy_prorated(row, power), axis=1
)

# ── 6. Summary ────────────────────────────────────────────────────────────────
cols = ["job_id", "start_time", "end_time", "num_nodes",
        "node_energy_kwh", "facility_energy_kwh_prorated"]
print(jobs[cols].dropna(subset=["node_energy_kwh"]).head(10))
print("\nEnergy stats (kWh):")
print(jobs["node_energy_kwh"].describe())
```
